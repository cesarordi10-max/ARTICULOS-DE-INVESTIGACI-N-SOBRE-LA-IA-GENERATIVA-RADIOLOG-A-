**Actividad grupal**

Diseña un sistema de laboratorio con objetos de equipos, muestras y técnicos. Cada equipo debe registrar
muestras procesadas y reportar su estado. Integra creación, interacción y consultas entre objetos.

**Entrega grupal**

Presenta la solución ejecutable, las pruebas de los casos principales y una explicación breve de las
responsabilidades de cada clase.

```python
class Muestra:
    def configurar(self, codigo: str, tipo: str):
        self.codigo = codigo
        self.tipo = tipo
        self.estado = "Pendiente"

    def __str__(self):
        return f"Muestra({self.codigo}, Tipo: {self.tipo}, Estado: {self.estado})"


class Equipo:
    def configurar(self, identificador: str, nombre: str):
        self.identificador = identificador
        self.nombre = nombre
        self.estado = "Operativo"
        self.muestras_procesadas = []

    def procesar_muestra(self, muestra: Muestra) -> bool:
        if self.estado != "Operativo":
            return False
        if muestra.estado == "Procesada":
            return False
        muestra.estado = "Procesada"
        self.muestras_procesadas.append(muestra)
        return True

    def reportar_estado(self):
        return f"Equipo {self.identificador} ({self.nombre}) - Estado: {self.estado} | Muestras procesadas: {len(self.muestras_procesadas)}"


class Tecnico:
    def configurar(self, id_tecnico: str, nombre: str):
        self.id_tecnico = id_tecnico
        self.nombre = nombre

    def asignar_procesamiento(self, equipo: Equipo, muestra: Muestra) -> bool:
        return equipo.procesar_muestra(muestra)


# --- PRUEBAS Y DEMOSTRACIÓN ---

# 1. Instanciación e inicialización de objetos
m1, m2, m3 = Muestra(), Muestra(), Muestra()
m1.configurar("M-001", "Sangre")
m2.configurar("M-002", "Orina")
m3.configurar("M-003", "Tejido")

eq1, eq2 = Equipo(), Equipo()
eq1.configurar("EQ-101", "Centrífuga")
eq2.configurar("EQ-102", "Espectrómetro")

tec1 = Tecnico()
tec1.configurar("T-01", "Carlos Pérez")

# 2. Interacciones y procesamiento
print("--- ESTADO INICIAL ---")
print(eq1.reportar_estado())
print(m1)

print("\n--- PROCESAMIENTO DE MUESTRAS ---")
exito1 = tec1.asignar_procesamiento(eq1, m1)
print(f"¿Muestra 1 procesada por {tec1.nombre}?: {exito1}")

exito2 = tec1.asignar_procesamiento(eq1, m2)
print(f"¿Muestra 2 procesada por {tec1.nombre}?: {exito2}")

# Intento de re-procesar una muestra ya procesada
exito3 = tec1.asignar_procesamiento(eq1, m1)
print(f"¿Re-procesamiento de Muestra 1 exitoso?: {exito3}")

# Simulación de un equipo fuera de servicio
eq2.estado = "En Mantenimiento"
exito4 = tec1.asignar_procesamiento(eq2, m3)
print(f"¿Muestra 3 procesada en equipo deshabilitado?: {exito4}")

print("\n--- ESTADO FINAL Y CONSULTAS ---")
print(eq1.reportar_estado())
print(eq2.reportar_estado())
print(m1)
print(m2)
print(m3)
```
